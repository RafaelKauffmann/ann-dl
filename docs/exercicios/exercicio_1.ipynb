```python
# B_2) 

#funcoes para calcular as razoes de separacao e os desvios padroes medios

# funcao recebe o desvio padrao de uma classe em formato de lista (dp_x, dp_y)
def desvio_padrao_medio(dp):
    dp_x = dp[0]
    dp_y = dp[1]
    return (dp_x + dp_y)/2

# funcao recebe as medias em formato de lista (u_x, u_y) e valores float para dp_medio
def separation_ratio(media_a,media_b,dp_medio_a,dp_medio_b):
    media_x = media_a[0]-media_b[0]
    media_y = media_a[1]-media_b[1]
    return (( media_x**2 + media_y**2 )**.5) / ( dp_medio_a + dp_medio_b )

#desvios padroes medios de cada classe

dp_medio_0 = desvio_padrao_medio(dp_0)
dp_medio_1 = desvio_padrao_medio(dp_1)
dp_medio_2 = desvio_padrao_medio(dp_2)
dp_medio_3 = desvio_padrao_medio(dp_3)

# razao de separacao entre as classes

classe_0_classe_1 = separation_ratio(media_0,media_1,dp_medio_0,dp_medio_1)
classe_0_classe_2 = separation_ratio(media_0,media_2,dp_medio_0,dp_medio_2)
classe_0_classe_3 = separation_ratio(media_0,media_3,dp_medio_0,dp_medio_3)
classe_1_classe_2 = separation_ratio(media_1,media_2,dp_medio_1,dp_medio_2)
classe_1_classe_3 = separation_ratio(media_1,media_3,dp_medio_1,dp_medio_3)
classe_2_classe_3 = separation_ratio(media_2,media_3,dp_medio_2,dp_medio_3)

```

```python
# Taxa de mistura

# funcao que calcula distancia entre pontos. Recebe duas listas em formato (x,y)
def distancia_pontos(ponto_a,ponto_b):

    a_x = ponto_a[0]
    a_y = ponto_a[1]
    b_x = ponto_b[0]
    b_y = ponto_b[1]

    return ((a_x - b_x)**2+(a_y - b_y)**2)**.5

# normalizando o banco de dados de classes de d["classe_a_s_b"] = [[0,1,...,n-1,n],[0,1,...,n-1,n]] para d["classe_a_s_b"] = [[0,0],[1,1],...,[n-1,n-1],[n,n]]

data_pontos = {}

for key in data.keys():
    data_pontos[f"{key}"] = []

for key in data.keys():
    for i in range (len(data[f"{key}"][0])):
        data_pontos[f"{key}"].append( [data[f"{key}"][0][i],data[f"{key}"][1][i]])

```

```python
# calculando taxa de mistura das classes para cada valor de s

# contador de pontos distantes dos centros da propria classe
falso_centro = 0

# DataFrame das taxas de mistura das classes para os valores de s
taxa_mistura_s = {}

for s_value in s:
    for i in range(len(classe_0[0])):

        ponto_classe_0 = data_pontos[f"classe_0_{s_value}"][i]
        ponto_classe_1 = data_pontos[f"classe_1_{s_value}"][i]
        ponto_classe_2 = data_pontos[f"classe_2_{s_value}"][i]
        ponto_classe_3 = data_pontos[f"classe_3_{s_value}"][i]

        dist_classe_0_media_0 = distancia_pontos(ponto_classe_0,media_0)
        dist_classe_0_media_1 = distancia_pontos(ponto_classe_0,media_1)
        dist_classe_0_media_2 = distancia_pontos(ponto_classe_0,media_2)
        dist_classe_0_media_3 = distancia_pontos(ponto_classe_0,media_3)

        dist_classe_1_media_0 = distancia_pontos(ponto_classe_1,media_0)
        dist_classe_1_media_1 = distancia_pontos(ponto_classe_1,media_1)
        dist_classe_1_media_2 = distancia_pontos(ponto_classe_1,media_2)
        dist_classe_1_media_3 = distancia_pontos(ponto_classe_1,media_3)

        dist_classe_2_media_0 = distancia_pontos(ponto_classe_2,media_0)
        dist_classe_2_media_1 = distancia_pontos(ponto_classe_2,media_1)
        dist_classe_2_media_2 = distancia_pontos(ponto_classe_2,media_2)
        dist_classe_2_media_3 = distancia_pontos(ponto_classe_2,media_3)

        dist_classe_3_media_0 = distancia_pontos(ponto_classe_3,media_0)
        dist_classe_3_media_1 = distancia_pontos(ponto_classe_3,media_1)
        dist_classe_3_media_2 = distancia_pontos(ponto_classe_3,media_2)
        dist_classe_3_media_3 = distancia_pontos(ponto_classe_3,media_3)

        distancias_classe_0 = [dist_classe_0_media_0, dist_classe_0_media_1, dist_classe_0_media_2, dist_classe_0_media_3]
        distancias_classe_1 = [dist_classe_1_media_0, dist_classe_1_media_1, dist_classe_1_media_2, dist_classe_1_media_3]
        distancias_classe_2 = [dist_classe_2_media_0, dist_classe_2_media_1, dist_classe_2_media_2, dist_classe_2_media_3]
        distancias_classe_3 = [dist_classe_3_media_0, dist_classe_3_media_1, dist_classe_3_media_2, dist_classe_3_media_3]

        if min(distancias_classe_0) != dist_classe_0_media_0:
            falso_centro += 1

        if min(distancias_classe_1) != dist_classe_1_media_1:
                    falso_centro += 1

        if min(distancias_classe_2) != dist_classe_2_media_2:
                    falso_centro += 1

        if min(distancias_classe_3) != dist_classe_3_media_3:
                    falso_centro += 1
    taxa_mistura_s[f"{s_value}"] = falso_centro/400
    falso_centro = 0


print(taxa_mistura_s)



```

### C - Análise

#### C1 — Sobreposição e separação linear

No conjunto de dados original, correspondente a (s = 1), observa-se uma sobreposição mais significativa entre as classes 0 e 1, resultado coerente com o fato de esse par apresentar a menor razão de separação, $r_{01}=1{,}325825$. Também existe uma aproximação entre as regiões ocupadas pelas classes 1 e 2, enquanto a classe 3 permanece mais isolada devido à maior distância de sua média em relação às demais.

Uma única fronteira linear não é suficiente para separar as quatro classes, pois uma reta divide o plano em apenas duas regiões. Um conjunto de fronteiras lineares pode criar várias regiões e separar corretamente grande parte dos pontos. Entretanto, essa separação não será perfeita nas regiões em que pontos de classes diferentes se sobrepõem, pois não existe uma fronteira capaz de distinguir corretamente todos os pontos que ocupam regiões semelhantes do espaço.

#### C2 — Esboço das fronteiras de decisão

As fronteiras da Figura 1, foram esboçadas aproximadamente entre as médias das classes vizinhas, procurando atravessar regiões com menor concentração de pontos. As linhas tracejadas representam possíveis fronteiras lineares que uma rede neural poderia combinar para formar diferentes regiões de decisão. Elas constituem apenas uma interpretação geométrica dos dados, pois nenhum modelo foi treinado nesta atividade.

### C3 — Espalhamento e região de erro

Conforme o fator de escala \(s\) aumenta, os desvios-padrão crescem enquanto as médias permanecem fixas. Consequentemente, as nuvens ocupam regiões maiores do plano e passam a apresentar mais sobreposição. Esse comportamento é confirmado pelo aumento da taxa de mistura, que passa de \(0{,}25\%\), em \(s = 0,5\), para \(7,25 \%\), em \(s = 1\), \(23,25 \%\), em \(s = 2\), e \(43,50 \%\), em \(s = 4\).

Com o aumento da sobreposição, também cresce a região na qual uma fronteira de decisão precisa atravessar áreas ocupadas por pontos de classes diferentes. Nessas regiões, mesmo a combinação de várias fronteiras lineares não consegue classificar todos os pontos corretamente. Portanto, quanto maior o espalhamento das nuvens, maior é a região de ambiguidade e maior tende a ser a quantidade de erros inevitáveis de classificação.
